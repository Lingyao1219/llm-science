```python
import os
import pandas as pd
from paper_processor import DocumentProcessor
from relevance_checker import RelevanceChecker
from affiliation_processor import AffiliationProcessor
from affiliation_matcher import AffiliationMatcher
from entropy_calculator import EntropyCalculator

import warnings
warnings.filterwarnings('ignore')
```

```python
types = ['preprint', 'article']
start_date = "2018-10-01"
end_date = "2024-09-30"

processor = DocumentProcessor()
checker = RelevanceChecker()
matcher = AffiliationMatcher()
calculator = EntropyCalculator()
```

### Processing LLM papers
#### 1. Cleaning papers

```python
df1 = processor.sample_documents('search-llm', 1)
print(f"The total number of downloaded papers: {len(df1)}")
df1 = processor.filter_by_date(df1, start_date, end_date)
print(f"The total number of papers after sorting date: {len(df1)}")
df1 = processor.filter_by_type(df1)
print(f"The total number of papers after filtering paper types: {len(df1)}")
df1 = processor.remove_exact_duplicates(df1)
print(f"The total number of papers after removing duplicates: {len(df1)}")
df1 = processor.remove_similar_documents(df1)
print(f"The total number of papers after potential duplicates: {len(df1)}")
```

#### 2. Checking relevance

```python
batch_size = 200
all_results = []
save_directory = "search-llm-relevance"
if not os.path.exists(save_directory):
    os.makedirs(save_directory)

for start_idx in range(0, len(df1), batch_size):
    batch_df = df1.iloc[start_idx:start_idx + batch_size]
    batch_number = start_idx // batch_size + 1
    batch_results = checker.check_papers(batch_df)
    batch_results.to_csv(f'{save_directory}/check_batch_{batch_number}.csv', index=False)
    all_results.append(batch_results)
    print(f"Batch {batch_number} processed and saved.")

df1 = pd.concat(all_results, ignore_index=True)
```

#### 3. Extracting department information

```python
save_directory = "llm-author-affiliations"

extractor = AffiliationProcessor(
    batch_size=200,
    save_directory=save_directory
)

affiliations1 = extractor.extract_affiliation(df1)
affiliations1 = extractor.prepare_prompt(affiliations1)
print(f"The total number of authors: {len(affiliations1)}")
extractor.process_batches(affiliations1)
```

```python
df1 = pd.read_csv('llm-cleaned-samples.csv')

folder_path1 = "llm-author-affiliations"

df1 = matcher.process_affiliation(df1)
affiliation_df1 = df1['affiliation_strings'].explode().value_counts().reset_index()
compile_df1 = matcher.compile_files(folder_path1)
compile_df1 = compile_df1.rename(columns={'affiliation_string': 'affiliation_strings'})
mapped_affiliation_df1 = matcher.create_affiliation_mapping(affiliation_df1, compile_df1)

df1 = matcher.process_departments(df1)
```

#### 4. Calculating Entropy

```python
df1 = calculator.calculate_all_entropies(df1)
df1.to_csv('llm-information-entropy.csv',index=False)
df1.head(5)
```
